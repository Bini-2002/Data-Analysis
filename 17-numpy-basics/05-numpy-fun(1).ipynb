```python
import numpy as np
```

```python
arr_a = np.array([1, 2, 3])
arr_a
```

```python
arr_b = np.array([[1], [2]])
arr_b
```

```python
matrix_c = np.array([[1, 2, 3], [4, 5, 6]])
matrix_c
```

  ## Broadcasting

```python
np.add(arr_a, matrix_c)
```

## Universal Functions

https://numpy.org/doc/stable/reference/ufuncs.html

## Type Casting

```python
np.add(arr_a, matrix_c, dtype=np.float64)
```

```python
np.add(arr_a, matrix_c, dtype=np.int64)
```

## Running over axis

```python
np.mean(matrix_c, axis=0)  # Column-wise mean , axis=0 : down the rows
```

```python
matrix_c
```

```python
np.mean(matrix_c, axis=1)  # Row-wise mean , axis=1 : across the columns
```
