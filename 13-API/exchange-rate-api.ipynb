```python
base_url = "https://api.exchangeratesapi.io/v1/latest"
```

```python
import requests
import os
```

```python
# Make a safe request: use API key from env if available, otherwise use free API
api_key = os.environ.get("EXCHANGE_RATES_API_KEY")
if api_key:
    params = {"access_key": api_key}
    url = base_url
else:
    # fallback free endpoint and params
    url = "https://api.exchangerate.host/latest"
    params = {"symbols": "USD,EUR,GBP"}
try:
    response = requests.get(url, params=params, timeout=10)
except requests.RequestException as e:
    print("Request failed:", e)
    raise
print("Requested URL:", response.url)
print("Status code:", response.status_code)
if not response.ok:
    try:
        print("Error JSON:", response.json())
    except Exception:
        print("Error text:", response.text)
    response.raise_for_status()
data = response.json()
# print a small summary of the returned data
print("Base:", data.get("base") or data.get("base_code") or "N/A")
print("Rates keys sample:", list((data.get("rates") or {}).keys())[:10])
```
