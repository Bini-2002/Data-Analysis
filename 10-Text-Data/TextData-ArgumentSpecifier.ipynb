An Argument Specifier

- is a combination of symbols that indicates the place within
the string where you'd like to insert information from a separate object(the argument)

- specifies the data type of that argument

Modulus Operator

- allows us to point to the variable from which we'd like 
  to obtain info and insert it into the original string


```python
product_category = ["A", "B"]
```

```python
print("This item has a product category of %s" % product_category[0])
```

```python
quantity = [500, 600]
print("The quantity ordered is %d available units" % quantity[0])
```

```python
stock_share_price = [23.45, 45.56]
print("The current stock price is $%.2f" % stock_share_price[0])
```

```python
i = 0
print(
    "Currently , we have %d units of category %s products in store"
    % (quantity[i], product_category[i])
)
```
