```python
a = (1, 2, 3, 4)
b = "abcde"
c = [10, 20, 30, 40, 50]
d = {"name": "Alice", "age": 25, "city": "New York"}
a
```

```python
for i in a:  # iterating through a tuple
    print(i, end=" ")  # prints 1 2 3 4
```

```python
for char in b:  # iterating through a string
    print(char, end=" ")  # prints a b c d e
```

```python
for num in c:  # iterating through a list
    print(num, end=" ")  # prints 10 20 30 40 50
```

```python
for key in d:  # iterating through a dictionary
    print(key, d[key])  # prints name Alice age 25 city New York
```

```python
for i in range(5):  # iterating through a range
    print(i, end=" ")  # prints 0 1 2 3 4
```

Nested Loops

```python
for i in range(2):
    for j in range(3):
        print(f"i={i}, j={j}")
```

```python
for i in ["Product A", "Product B"]:
    for j in range(5):
        print(f"{i} - {j}")
```

```python
products = ["Product A", "Product B"]
exp_sales = [100, 150, 200, 250, 300]
```

```python
for i in products:
    for j in exp_sales:
        print([i, j])
```

Triple Nested Loops

```python
products = ["Product A", "Product B"]
exp_sales = [100, 150, 200, 250, 300]
```

```python
time_horizon = (1, 3, 12)
```

```python
for prod in products:
    for sale in exp_sales:
        for t_hor in time_horizon:
            print([prod, sale * t_hor])
```

```python
for prod in products:
    for sale in exp_sales:
        for t_hor in time_horizon:
            print(
                "Expected sales for a period of {0} month(s) for {1}: ${sales}".format(
                    t_hor, prod, sales=sale * t_hor
                )
            )
```
