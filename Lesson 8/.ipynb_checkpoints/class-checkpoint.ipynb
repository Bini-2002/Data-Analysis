Using An Index in Pandas

```python
prices_per_category = {'Product A' : 30000 , 'Product B': 20000 , 'Product C': 15000}

prices_per_category = pd.Series(prices_per_category)

type(prices_per_category)
```

```python
prices_per_category.index 
```

```python
type(prices_per_category.index)
```

```python
series_b = pd.Series([10,20,30] , index = [1,2,3])
series_b
```

```python
series_b = pd.Series([10,20,30,40] , index = ["1","2","3","4"])
series_b
```

```python
series_b[1]
```

```python
series_b["1"]
```
