```python
import pandas as pd
```

```python
data = {'ProductName' : ['Product A','Product B','Product C'] , 'ProductPrice': [22500 , 16600 , 12500]}
```

```python
data = {'ProductName' : ['Product A','Product B','Product C'] , 'ProductPrice': [22500 , 16600 , 12500]}
df = pd.DataFrame(data , index = ['A','B','C'])
df
```
